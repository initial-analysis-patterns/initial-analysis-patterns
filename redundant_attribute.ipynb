## Redundant Attribute

**Modification Objective:** Remove an attribute that has been identified as carrying no relevant information.

**Motivation:** Attributes identified as redundant encode information already represented by, or functionally dependent on, other attributes in the event log. This includes attributes in a 1:1 relationship as well as attributes that are functionally dependent on another attribute, e.g., through an n:1 relationship. Attributes identified as constant throughout the event log encode no distinguishing information between events. Retaining such attributes can unnecessarily increase the complexity of the event data and may lead analysts to treat equivalent, dependent, or constant information as distinct dimensions in subsequent analyses. Removing such attributes or storing dependent information separately, on the other hand, can provide a more concise representation.

**Precondition:** An attribute has been identified as uninformative.

**Approach:** Remove a selected uninformative attribute while retaining, in the case of redundant attributes, the attribute that provides its equivalent information.

**Output:** An event log in which the selected uninformative attribute has been removed.

```python
import pandas as pd
import pm4py

# --- Configuration -----------------------------------------------------------
LOG_PATH = "../../data/Road_Traffic_Fine_Management_Process.xes"

DROP_ATTRIBUTE_NAME = "lifecycle:transition"
```

```python
event_log = pm4py.read_xes(LOG_PATH)

display(event_log.head())
```

## Pattern execution

```python
modified_event_log = event_log.copy()

modified_event_log = modified_event_log.drop(columns=DROP_ATTRIBUTE_NAME)

display(modified_event_log.head())
```
