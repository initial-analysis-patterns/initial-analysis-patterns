# Event Ordering
**Information Need:** Verify if the events within each case and overall in the log are ordered according to the timestamp attribute, and whether the timestamps establish a total or partial ordering of events; where required, reorder events according to their timestamps..

**Motivation:** Several analyses and transformations rely on the sequence in which events occur within a case. If events are not stored in chronological order, sequential operations may produce incorrect results even when valid timestamps are available. Reordering events according to their timestamps establishes a sequence consistent with the available temporal information. However, when multiple events share the same timestamp, their relative order cannot be determined from timestamps alone.

**Precondition**: If reordering is performed and relevant, a rule specifying how to handle partial order is specified.

**Approach:** Assess whether the recorded sequence of events is consistent with their timestamps and whether timestamp ties leave the relative order of events undetermined. If required, reorder events according to their timestamps while preserving their association with their respective cases.

**Output:** The indication, at both case- and log-level, of whether events are chronologically ordered and whether their timestamps induce a total or partial order, together with the cases and events for which ordering violations or timestamp ties occur; if reordering is performed, an event log in which events are ordered chronologically according to the specified timestamp attribute.

```python
import pm4py
import pandas as pd 

# --- Configuration -----------------------------------------------------------

LOG_PATH = "../../data/BPIC15_1.xes"

CASE_ID = "case:concept:name"
TIMESTAMP = "time:timestamp"
```

```python
event_log = pm4py.read_xes(LOG_PATH)

display(event_log.head())
```

### 1. Case-wise Event Ordering

```python
#assert event_log.groupby(CASE_ID)[TIMESTAMP].is_monotonic_increasing.all()
display(event_log.groupby(CASE_ID)[TIMESTAMP].is_monotonic_increasing.all())
```

### 2. Total Event Ordering

```python
#assert event_log[TIMESTAMP].is_monotonic_increasing
display(event_log[TIMESTAMP].is_monotonic_increasing)
```

```python
modified_event_log = event_log.copy()

#this might take a while
modified_event_log['original_order'] = range(len(event_log))
modified_event_log = modified_event_log.sort_values(by=[CASE_ID, TIMESTAMP])

#check if ordered now
assert modified_event_log.groupby(CASE_ID)[TIMESTAMP].is_monotonic_increasing.all()
display(modified_event_log.groupby(CASE_ID)[TIMESTAMP].is_monotonic_increasing.all())
```

```python
modified_event_log = event_log.copy()

modified_event_log['original_order'] = range(len(event_log))
modified_event_log = modified_event_log.sort_values(by=[TIMESTAMP, 'original_order'])

#check if ordered now
assert modified_event_log[TIMESTAMP].is_monotonic_increasing
display(modified_event_log[TIMESTAMP].is_monotonic_increasing)
```
