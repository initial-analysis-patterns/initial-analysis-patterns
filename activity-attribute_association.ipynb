## Activity–Attribute Association

**Information Need:** Assess whether an attribute recorded for an activity is intrinsically associated with that activity.

**Motivation:** Event logs may associate attributes with activities even though their values are determined by other events. Such associations can lead analysts to incorrectly interpret an attribute as describing the activity itself. Identified mismatches may motivate subsequent transformations of the event log, such as adding events to make implicitly represented activities explicit, removing events that do not represent distinct activity occurrences, or relocating information to the activity with which it is semantically associated.

**Precondition:** The analyst has an understanding of the semantics of the activity types and attributes.

**Approach:** Compare where the selected attribute is populated in the event log with where, according to its semantics and the process context, the information represented by the attribute is expected to become available. If required, transform the event log accordingly, for example by adding or removing events or changing the association of information with activity types.

**Output:** An assessment of whether the selected attribute is semantically associated with the activities for which it is recorded, including identified mismatches between its observed and expected association with activity types; if a transformation is performed, an event log in which the identified association is represented accordingly.



For example, consider executing patterns such as `Activity Schema`, `Activity Schema Overlap`, and `Attribute Value Characterization` to assess if a relationship between an activity and attribute is meaningful within the process domain.
